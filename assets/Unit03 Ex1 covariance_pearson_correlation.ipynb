### Unit 3 - Correlation and Regression experimentation

In this file, we're using synthetic data that is generated randomly using Python.
The initial generated data creates a positive correlation between two variables.
The assignment is to manipulate the data and observe changes in the correlation.

```python
from numpy import mean
from numpy import std
from numpy import cov
from numpy.random import randn
from numpy.random import seed
from matplotlib import pyplot as plt
import seaborn as sns

from scipy.stats import pearsonr
```

```python
# calculate the Pearson's correlation between two variables

# seed random number generator
seed(1)

# prepare data
data1 = 20 * randn(1000) + 100
data2 = data1 + (10 * randn(1000) + 50)

# calculate covariance matrix
covariance = cov(data1, data2)

# calculate Pearson's correlation
corr, _ = pearsonr(data1, data2)

# plot
plt.scatter(data1, data2)
plt.show()

# summarize
print('data1: mean=%.3f stdv=%.3f' % (mean(data1), std(data1)))
print('data2: mean=%.3f stdv=%.3f' % (mean(data2), std(data2)))
print('Covariance: %.3f' % covariance[0][1])
print('Pearsons correlation: %.3f' % corr)
```

#### Manipulation 1: Creating a negative correlation
By creating a new variable, data3, with a negative relationship to data1, while using the same level of noise and randomness, we maintain a similarly strong correlation, but it becomes negative. This is reflected by the decreasing linear trend in the generated plot.

In addition, the mean of data3 is different. By adding a negative sign to data1 in the assignment, we've essentially moved from:

100+50=150

to:

−100+50=−50

Lastly, we see a slightly different magnitude of Pearson's correlation (−0.901 compared with 0.888). This is likely because we generated a new set of random values for data3, so the random noise behaves slightly differently.

```python
data3 = -data1 + (10 * randn(1000) + 50)

# calculate covariance matrix
covariance = cov(data1, data3)

# calculate Pearson's correlation
corr, _ = pearsonr(data1, data3)

# plot
plt.scatter(data1, data3)
plt.show()

# summarize
print('data1: mean=%.3f stdv=%.3f' % (mean(data1), std(data1)))
print('data3: mean=%.3f stdv=%.3f' % (mean(data3), std(data3)))
print('Covariance: %.3f' % covariance[0][1])
print('Pearsons correlation: %.3f' % corr)
```

### Manipulation 2 - Increase random noise
By keeping everything else the same but increasing the noise for a new variable, data4, we make the linear relationship between data1 and data4 much weaker and less apparent in the data.

The noise is increased by changing:

data4 = data1 + (10 * randn(1000) + 50)

To:

data4 = data1 + (100 * randn(1000) + 50)

Pearson's correlation consequently decreases significantly in strength (from 0.88 to 0.152). As expected, the standard deviation of data4 also increases substantially (from 22.35 to 102.86), because the values are now much more widely dispersed.

```python
data4 = data1 + (100 * randn(1000) + 50)

# calculate covariance matrix
covariance = cov(data1, data4)

# calculate Pearson's correlation
corr, _ = pearsonr(data1, data4)

# plot
plt.scatter(data1, data4)
plt.show()

# summarize
print('data1: mean=%.3f stdv=%.3f' % (mean(data1), std(data1)))
print('data4: mean=%.3f stdv=%.3f' % (mean(data4), std(data4)))
print('Covariance: %.3f' % covariance[0][1])
print('Pearsons correlation: %.3f' % corr)
```
